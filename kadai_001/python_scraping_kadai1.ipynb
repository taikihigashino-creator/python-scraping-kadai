```python
import requests
from bs4 import BeautifulSoup

# Yahoo!ニュースの記事URL
url = 'https://news.yahoo.co.jp/articles/db843537dbddc331ba5b75a4d2001284a084f7be'

# HTMLを取得
response = requests.get(url)

# HTMLを解析
soup = BeautifulSoup(response.text, 'html.parser')

# 記事本文を取得
body_elements = soup.select('article#uamods > div > div > p')

# 本文を表示
for body_element in body_elements:
    print(body_element.text)
```
